#### Import packages

```python
import altair as alt
import pandas as pd
```

```python
url = 'https://gist.githubusercontent.com/officeofjane/2c3ed88c4be050d92765de912d71b7c4/raw/8abc5e19a5fac2ec63e81891d9bfdebe2d55f92c/publication-grids.csv'
df = pd.read_csv(url)
```

```python
# filter the data to include only New York Times 2
df = df[df['publication'] == 'New York Times 1']
```

#### Trying with my own data

```python
csv = pd.read_csv("data/50 states and abbreviations - Sheet1-2.csv")
```

```python
csv.head()
```

```python
df = df.merge(csv, left_on='code', right_on='Abbreviation', how='left')
#_states_dur = pd.read_csv(url)
```

```python
df.head()
```

```python
# import altair as alt

heatmap = alt.Chart(df).mark_rect(stroke=None, cornerRadius=5).encode(
    x=alt.X('col:O', axis=None),
    y=alt.Y('row:O', axis=None),
    color=alt.Color('days:Q', scale=alt.Scale(scheme='blues'), legend=alt.Legend(title='Days')),
    tooltip=['code', 'days']
).properties(
    width=600,
    height=400,
    title='State-Level Days Heatmap'
)
```

```python
heatmap #.display()
```

```python
text = heatmap.mark_text(baseline='middle', align='center', fill='black', fontSize=12).encode(
    text='code:O'
)
```

```python
text
```

```python
chart = (heatmap + text).properties(width=350, height=250).configure_view(strokeOpacity=0)
```

```python
# display the chart
chart
```

```python
# text = heatmap.mark_text(baseline='middle', align='center', fill='black', fontSize=12).encode(
#     text='code:O'
# )

# Text overlay layer
labels = heatmap.mark_text(
    align='center',
    baseline='middle',
    # color='white',
    fill='white',
    fontSize=14,
    fontWeight='bold'
).encode(
    text='code:O'
)
```

```python
labels
```

```python
# Combine layers
chart = (heatmap + labels).properties(
    width=600,
    height=400,
    title='State-Level Days Heatmap (Reversed Magma + Labels)'
)

chart.display()
```

#### Working on the Labels

```python
# Text shadow layer
# text_outline = heatmap.mark_text(
#     align='center',
#     baseline='middle',
#     # color='black',
#     fill='black',
#     opacity=0.4,
#     fontSize=14,
#     fontWeight='bold'
# ).encode(
#     text='code:O'
# )

# Create fake text outline using small offsets in 4 directions
offsets = [(-0.5, 0), (0.5, 0), (0, -0.5), (0, 0.5)]

shadow_layers = [
    heatmap.mark_text(
        align='center',
        baseline='middle',
        fill='black',
        opacity=0.3,
        fontSize=12,
        fontWeight='bold',
        dx=dx,
        dy=dy
    ).encode(
        text='code:O'
    )
    for dx, dy in offsets
]

# Foreground white text layer
text_foreground = heatmap.mark_text(
    align='center',
    baseline='middle',
    # color='white',
    fill='white',
    fontSize=12,
    # fontWeight='bold'
).encode(
    text='code:O'
)
```

```python
# chart = (heatmap + text_outline + text_foreground).properties(
#     width=600,
#     height=400,
#     title='State-Level Days Heatmap (Reversed Magma + Labels)'
# )

# Combine all layers
chart = alt.layer(
    heatmap,
    *shadow_layers,
    text_foreground
# ).resolve_scale(
#     color='independent'
).properties(
    width=600,
    height=400,
    # title='State-Level Days Heatmap (Log Color, Label Outline)'
    title="Where I've spent my days"
)

chart.display()
```

```python
base = alt.Chart(df).encode(
    x=alt.X('col:O', axis=None),
    y=alt.Y('row:O', axis=None)
)

# Rect layer (heatmap)
heatmap = base.mark_rect(stroke=None, cornerRadius=5).encode(
    color=alt.Color(
        'days_log:Q',
        scale=alt.Scale(type='log', scheme='magma', reverse=True),
        legend=alt.Legend(title='Days (log)')
    )
)

# Text overlay layer
labels = heatmap.mark_text(
    align='center',
    baseline='middle',
    color='white',
    fontSize=12,
    fontWeight='bold'
).encode(
    text='days:Q'
)

# Combine layers
chart = (heatmap + labels).properties(
    width=600,
    height=400,
    title='State-Level Days Heatmap (Reversed Magma + Labels)'
)

chart.display()
```
